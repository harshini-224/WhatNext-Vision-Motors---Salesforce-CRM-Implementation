```python
import pandas as pd
import random
from faker import Faker

fake = Faker()

# 1. Vehicle__c
vehicle_models = ['Sedan', 'SUV', 'EV', 'Truck']
vehicle_status = ['Available', 'Out of Stock', 'Discontinued']
vehicles = []
for i in range(20):
    vehicles.append({
        "Vehicle_Name__c": f"{fake.color_name()} {fake.word().capitalize()}",
        "Vehicle_Model__c": random.choice(vehicle_models),
        "Stock_Quantity__c": random.randint(0, 50),
        "Price__c": round(random.uniform(20000, 70000), 2),
        "Dealer__c": f"Dealer_{random.randint(1, 10)}",
        "Status__c": random.choice(vehicle_status)
    })

# 2. Vehicle_Dealer__c
dealers = []
for i in range(10):
    dealers.append({
        "Dealer_Name__c": fake.company(),
        "Dealer_Location__c": fake.city(),
        "Dealer_Code__c": f"D{i+1:03d}",
        "Phone__c": fake.phone_number(),
        "Email__c": fake.company_email()
    })

# 3. Vehicle_Customer__c
customers = []
for i in range(20):
    customers.append({
        "Customer_Name__c": fake.name(),
        "Email__c": fake.email(),
        "Phone__c": fake.phone_number(),
        "Address__c": fake.address().replace('\n', ', '),
        "Preferred_Vehicle_Type__c": random.choice(vehicle_models)
    })

# 4. Vehicle_Order__c
order_status = ['Pending', 'Confirmed', 'Delivered', 'Canceled']
orders = []
for i in range(20):
    orders.append({
        "Customer__c": f"Customer_{random.randint(1, 20)}",
        "Vehicle__c": f"Vehicle_{random.randint(1, 20)}",
        "Order_Date__c": fake.date_between(start_date='-1y', end_date='today'),
        "Status__c": random.choice(order_status)
    })

# 5. Vehicle_Test_Drive__c
test_drive_status = ['Scheduled', 'Completed', 'Canceled']
test_drives = []
for i in range(20):
    test_drives.append({
        "Customer__c": f"Customer_{random.randint(1, 20)}",
        "Vehicle__c": f"Vehicle_{random.randint(1, 20)}",
        "Test_Drive_Date__c": fake.date_between(start_date='-6m', end_date='today'),
        "Status__c": random.choice(test_drive_status)
    })

# 6. Vehicle_Service_Request__c
service_status = ['Requested', 'In Progress', 'Completed']
services = []
for i in range(20):
    services.append({
        "Customer__c": f"Customer_{random.randint(1, 20)}",
        "Vehicle__c": f"Vehicle_{random.randint(1, 20)}",
        "Service_Date__c": fake.date_between(start_date='-1y', end_date='today'),
        "Issue_Description__c": fake.sentence(),
        "Status__c": random.choice(service_status)
    })

# Create Excel file
with pd.ExcelWriter("Vehicle_System_Sample_Data.xlsx") as writer:
    pd.DataFrame(vehicles).to_excel(writer, sheet_name="Vehicle__c", index=False)
    pd.DataFrame(dealers).to_excel(writer, sheet_name="Vehicle_Dealer__c", index=False)
    pd.DataFrame(customers).to_excel(writer, sheet_name="Vehicle_Customer__c", index=False)
    pd.DataFrame(orders).to_excel(writer, sheet_name="Vehicle_Order__c", index=False)
    pd.DataFrame(test_drives).to_excel(writer, sheet_name="Vehicle_Test_Drive__c", index=False)
    pd.DataFrame(services).to_excel(writer, sheet_name="Vehicle_Service_Request__c", index=False)
```

```python
import pandas as pd

# Load the Excel file
excel_file = pd.ExcelFile("Vehicle_System_Sample_Data.xlsx")

# Loop through all sheet names and save each as a CSV
for sheet_name in excel_file.sheet_names:
    df = pd.read_excel(excel_file, sheet_name=sheet_name)
    csv_filename = f"{sheet_name}.csv"
    df.to_csv(csv_filename, index=False)
    print(f"Saved: {csv_filename}")
```
